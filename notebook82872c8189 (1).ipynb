```python
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS
text="This is an example sentence demonstrating stopwords removal and its impact on Visualization"
words=text.lower().split()
filtered_words=[word for word in words if word not in ENGLISH_STOP_WORDS]
words_count=Counter(words)
filtered_count=Counter(filtered_words)
plt.figure(figsize=(10,5))
plt.bar(words_count.keys(),words_count.values(),alpha=0.5,label="With Sropwords")
plt.bar(filtered_count.keys(),filtered_count.values(),alpha=0.5,label="Without Stopwords")
plt.xticks(rotation=45)
plt.legend()
plt.title("Word Frequency Comparision With VS Without StopWords")
plt.show()
```
